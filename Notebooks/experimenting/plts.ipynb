```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
x = np.arange(0, 10, 1)
y = 1 + 0.5*x + np.random.normal(loc=0.0, scale=0.5, size=10)

fig, ax = plt.subplots()
ax.scatter(x, y)
plt.xlim([0, 10])
plt.ylim([0, 10])
plt.show()
```

```python
def f(x, theta):
    return x**2 + theta

x = np.arange(0, 10)
thetas = np.array([1, 5, 10, 50, 100])
ys = np.zeros((thetas.shape[0], 10))
fig, ax = plt.subplots()

for c_theta in range(thetas.shape[0]):
    theta = thetas[c_theta]
    for i in range(10):
        ys[c_theta, i] = f(i, theta)
    ax.plot(x, ys[c_theta, :])

plt.xlim([0,10])
plt.show()

```
